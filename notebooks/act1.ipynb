```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

x = np.random.normal(10,2,200)
error = np.random.normal(0, 1.5, 200)
 
b0_real = 3.0
b1_real = 2.0

y = b1_real * x + b0_real + error
```

```python
plt.figure(figsize=(11, 6))
plt.scatter(x, y)
plt.xlabel('x (Variable predictora)')
plt.ylabel('y (Variable respuesta)')
plt.title('Datos Sintéticos: Relación Lineal con Ruido')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
x_mean = x.mean()
y_mean= y.mean()
x_mean, y_mean
```

```python
sxx = sum((x - x_mean)**2)
sxy = sum((x - x_mean) * (y - y_mean))
sxx, sxy
```

```python
b1 = sxy / sxx
b0 = y_mean - b1 * x_mean
b1, b0
```

```python
y_pred = b1 * x + b0
residuos = y - y_pred
y_pred, residuos;
```

```python
sse = sum(residuos**2)
ssr = sum((y_pred - y_mean)**2)
sst = sum((y - y_mean)**2)
sse, ssr, sst
```

```python
r2 = 1 - (sse / sst)
r2
```

```python
mse = np.mean(residuos**2)
rmse = np.sqrt(mse)
rmse
```

```python
plt.figure(figsize=(11, 6))
plt.scatter(x, y)
plt.plot(x, y_pred, color='red', label='Regresión ajustada')
plt.plot(x, b1_real * x + b0_real, color='green', label='Relación real (sin ruido)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.title(f"Regresión Lineal Simple (R² = {r2:.2f})")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(11, 6))
plt.scatter(y_pred, residuos, color='green')
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel('Valores ajustados')
plt.ylabel('Residuos')
plt.title('Grafica de residuos')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
print(f"Ecuacion del modelo")
print(f"y = {b0:.4f} + {b1:.4f}x")
print(f"Real: y = {b0_real} + {b1_real}x")


print(f"Metricas encontradas")
print(f"R² = {r2:.4f} - {r2*100:.2f}% de variabilidad explicada")

print(f"SSE = {sse:.4f}")
print(f"SSR = {ssr:.4f}")
print(f"SST = {sst:.4f}")
print(f"MSE = {mse:.4f}")
print(f"RMSE = {rmse:.4f} - Error promedio de {rmse:.2f} ")

print(f"Comparacion con los valores reales")
print(f"Pendiente(β₁):  {b1:.4f}  (Real: {b1_real})  - Diferencia: {abs(b1 - b1_real):.4f}")
print(f"Intercepto(β₀): {b0:.4f}  (Real: {b0_real})  - Diferencia: {abs(b0 - b0_real):.4f}")
```
